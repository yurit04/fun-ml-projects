# Kernel Linear Regression example

```python
# import numpy as np
import matplotlib.pyplot as plt
from sklearn.kernel_ridge import KernelRidge

# Generate some synthetic data
rng = np.random.RandomState(42)
X = 5 * rng.rand(100, 1)
y = np.sin(X).ravel()
# Add some noise to the data
y[::5] += 3 * (0.5 - rng.rand(X.shape[0] // 5))

# Initialize the Kernel Ridge Regression model
# We use an RBF (Radial Basis Function) kernel, which is common for non-linear relationships
# alpha is the regularization strength (similar to Ridge Regression)
# gamma is the kernel coefficient for 'rbf', 'poly', and 'sigmoid' kernels
model = KernelRidge(alpha=1.0, kernel='rbf', gamma=0.1)

# Fit the model to the data
model.fit(X, y)

# Generate data for plotting the regression line
X_plot = np.linspace(0, 5, 1000)[:, None]
y_plot = model.predict(X_plot)

# Plot the original data and the fitted regression line
plt.figure(figsize=(8, 6))
plt.scatter(X, y, color='black', label='Data')
plt.plot(X_plot, y_plot, color='red', label='Kernel Ridge Regression Prediction')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Kernel Ridge Regression with RBF Kernel')
plt.legend()
plt.grid(True)
plt.show()
```
